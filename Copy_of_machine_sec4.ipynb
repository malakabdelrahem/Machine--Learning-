```python
import numpy as np

def softmax(z):
    return np.exp(z) / np.sum(np.exp(z))

print(np.round(softmax(np.array([2.0, 1.0, 0.1])), 3))
```

```python
import numpy as np
from sklearn.linear_model import Ridge, Lasso

rng = np.random.RandomState(0)
X = rng.randn(50,5)
y = X @ np.array([3.,0,0,2,0]) + 0.1*rng.randn(50)

print("Lasso:")
print(np.round(Lasso(alpha=0.1).fit(X,y).coef_,2))

print("Ridge:")
print(np.round(Ridge(alpha=1.0).fit(X,y).coef_,2))
```

```python
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Ridge, Lasso

data = load_diabetes()

X = data.data
y = data.target

print("Linear Regression:")
linear = LinearRegression()
linear.fit(X,y)
print(np.round(linear.coef_,2))


feature_names = np.array(data.feature_names)

for alpha in [0.01, 0.1, 1.0, 10.0]:

    print("\nAlpha:", alpha)

    ridge = Ridge(alpha=alpha)
    ridge.fit(X,y)
    print("Ridge:", np.round(ridge.coef_,2))

    lasso = Lasso(alpha=alpha)
    lasso.fit(X,y)
    print("Lasso:", np.round(lasso.coef_,2))

    print("Zero Features:")
    print(feature_names[lasso.coef_ == 0])
```
